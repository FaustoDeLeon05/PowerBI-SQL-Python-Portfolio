```python
import pandas as pd
import sqlalchemy as sa

# 1. Ruta de tu archivo Excel
excel_path = r"C:\Users\faust\Desktop\Personal\Cursos\Curso PBI\Proyectos\Dataset - Prueba.xlsx"

# Leer las hojas del Excel
df_ventas = pd.read_excel(excel_path, sheet_name="Ventas")
df_gastos = pd.read_excel(excel_path, sheet_name="Gastos")

print("--- Vista previa de Ventas ---")
display(df_ventas.head(3))

print("--- Vista previa de Gastos ---")
display(df_gastos.head(3))
```

```python
# Configuración del servidor SQL local
server = 'XAVIER-PC'
database = 'GlobalSalesDB'

# Crear el motor de conexión usando autenticación de Windows
connection_string = f"mssql+pyodbc://@{server}/{database}?driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes"
engine = sa.create_engine(connection_string)

# Guardar los DataFrames de Pandas como tablas relacionales en SQL Server
df_ventas.to_sql('Fact_Ventas', con=engine, if_exists='replace', index=False)
df_gastos.to_sql('Dim_Gastos', con=engine, if_exists='replace', index=False)

print("¡Tablas 'Fact_Ventas' y 'Dim_Gastos' creadas y cargadas con éxito en SQL Server!")
```
